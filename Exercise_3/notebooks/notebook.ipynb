# Exercise 3: Reproducible Workflow  
**Course:** BERN02 – Reproducible Data Science and Statistical Learning    
**Objective:** Demonstrating a reproducible data analysis workflow for parsing and processing fungal protein sequences     
**Dataset:** `Fet3_6Genomes.fasta` — FASTA file containing iron multicopper oxidase (Fet3-like) protein sequences across fungal species. This file was retrieved from privious course material.  

```python
import pandas as pd
import matplotlib.pyplot as plt
from Bio import SeqIO
```

```python
# Extract records into a list of dictionaries
data = [{"ID": record.id, "Description": record.description, "Sequence": str(record.seq), "Length": len(record.seq),} for record in SeqIO.parse("../data/Fet3_6Genomes.fasta", "fasta")]

# Convert the list into a pandas DataFrame
df = pd.DataFrame(data)

# Display the first few rows
df.head()
```

```python
# Plot sequence lengths
plt.figure(figsize=(8, 10))
plt.barh(df["ID"], df["Length"], color="skyblue")
plt.xlabel("Amino Acid Length")
plt.ylabel("Fungal Protein ID")
plt.title("Fet3-like Protein Sequence Length Comparison")
plt.gca().invert_yaxis()
plt.show()
```

## FAIR Data Principles Reflection

This workflow was constructed to align with the **FAIR** principles for open and reproducible science:

* **Findable:** The dataset (`Fet3_6Genomes.fasta`) is clearly named, stored in a structured directory (`data/`), and documented in this notebook with metadata describing its biological context (Fet3-like multicopper oxidases across 6 fungal genomes).
* **Accessible:** All data and workflow code are stored in an open GitHub repository, allowing anyone to clone and view the project using standard version control.
* **Interoperable:** The sequence data uses the standard FASTA format, an open and universally recognized file standard in bioinformatics. Code operations rely on standard open-source Python tools (`Biopython`, `pandas`).
* **Reusable:** The repository includes a `LICENSE` file granting explicit legal permission for reuse, alongside clear documentation of dependencies to ensure execution across different local computational environments.
